```python
import json
```

```python
def load_data(filename):
    with open(filename,"r") as f:
        data =json.load(f)

    return data
```

```python
data =load_data("store_data.json")
```

```python
def clean_data(data):
    text_to_num={"one":1 ,"two":2 ,"three":3 ,"four":4 ,"five":5 }
    cleaned_data=[]
    unique_users=set()
    for user in data:
        ratings=str(user["rating"])
        if (ratings in text_to_num):
            ratings=text_to_num[ratings]
        user["rating"]=ratings
        r_age=user.get("age")
        if (r_age == None):
            user["age"] = None
    #deduplication
        if(user["name"] in unique_users):
                continue
            
        unique_users.add(user["name"])
        cleaned_data.append(user)

    return cleaned_data
```

```python
clean_data(data)
```

# meaning ful insight

```python
def get_ratings(data):
    #avg rating
    tot_rating=0
    for user in data:
        tot_rating+= float(user["rating"])

    print(f"avg rating={tot_rating/len(data)}")
    #poor ratings
    poor_rating=0
    for user in data:
        if (float(user["rating"]) <=3):
            poor_rating+=1
    print(f"poor rating={poor_rating/len(data)*100}%")
```

```python
get_ratings(data)
```

# recommendation

```python
def get_recom(data):
    recom=[]

    for user in data:
        cur_recom={}
        cur_recom["name"]= user["name"]
        if (float(user["rating"]) <=4):
            cur_recom["brand"]="apple"
        else:
            cur_recom["brand"]="samsung"
        recom.append(cur_recom)
    return recom
```

```python
get_recom(data)
```
