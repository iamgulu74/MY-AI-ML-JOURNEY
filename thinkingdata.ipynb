# DUPLICATE DATA

```python
def clean_data(data):
    text_to_num={"one":1 ,"two":2 ,"three":3 ,"four":4 ,"five":5 }
    cleaned_data=[]
    unique_users=set()
    for user in data:
        ratings=str(user["rating"])
        if (ratings in text_to_num):
            ratings=text_to_num[ratings]
        user["rating"]=ratings
        r_age=user.get("age")
        if (r_age == None):
            user["age"] = None
    #deduplication
        if(user["name"] in unique_users):
                continue
            
        unique_users.add(user["name"])
        cleaned_data.append(user)

    return cleaned_data
```

```python
clean_data(data)
```
