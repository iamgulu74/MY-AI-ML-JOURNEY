```python
import numpy as np
```

```python
arr=np.array([[1,2,3],[4,5,6]])
reshape=arr.reshape((3,2))
print(reshape,arr.reshape)
print(reshape.shape)
```

```python
#indexing
arr=np.array([[1,2,3],[4,5,6]])
print(arr[0][1])
#fancy indexing
idx=[0,1,1]
print(arr[idx])
```

```python
#slicing
arr=np.array([[1,2,3],[4,5,6]])
print(arr[0:2])
```
